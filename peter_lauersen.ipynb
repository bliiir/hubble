```python

    import numpy as np
    from numpy import log10,sqrt
    import matplotlib.pyplot as plt
    from astropy import units as u
    from astropy.cosmology import Planck15
    import astropy.constants as cc
    from scipy.optimize import curve_fit

    def lin(x,a):
        return a * x

    z,dL = np.loadtxt('laura.dat',unpack=True)      #Read data
    v     = cc.c * z                                #Convert to velocity
    #print('v    =', v.to(u.km/u.s))
    dL    = dL * u.Mpc                              #Use units
    #print('dL   =', dL)
    d     = dL / (1+z)                              #Convert to physical distance
    #print('d    =', d)

    dax   = np.linspace(0,110) * u.Mpc              #Define axis for the plot
    H0    = Planck15.H0 # 70 * u.km/u.s / u.Mpc     #H0 from Planck data
    vHub  = H0 * dax                                #Velocities for Planck data

    coeff,cov = curve_fit(lin,                              #Perform fit using a linear relation
                          (d/u.Mpc).decompose().value,      #curve_fit doens't handle unit, so remove
                          (v/(u.km/u.s)).decompose().value) #
    H0fit = coeff[0] * u.km/u.s / u.Mpc                     #...and add again
    print('H0fit    =', H0fit)
    vfit  = H0fit * dax                             #Fitted velocity

    plt.clf()
    #plt.ylim([-20,-18.5])
    #plt.errorbar(z,d, xerr=sigz, yerr=[sigMBlo,sigMBhi], fmt='ro',mec='r')
    plt.scatter(d.to(u.Mpc),    v.to(u.km/u.s), label='Laura\'s data')
    plt.plot( dax.to(u.Mpc), vHub.to(u.km/u.s), label='Planck (2016)')
    plt.plot( dax.to(u.Mpc), vfit.to(u.km/u.s), label='Fit to data')
    plt.xlabel(r'$d_\mathrm{L}\,/\,\mathrm{Mpc}$')
    plt.ylabel(r'$v\,/\,\mathrm{km}\,\mathrm{s}^{-1}$')
    plt.legend()
```
